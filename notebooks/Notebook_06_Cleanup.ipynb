# Student 360 Workshop - Cleanup
## Notebook 06 (Clean up)

_Icons: 🛠️ Action  📌 Note  🔹 Info  🎓 Higher-ed context  ✅ Best practice_

---
### What this notebook does
1. 🛠️ **Unpublishes** the lab's agents from the shared CoWork object
2. 🛠️ Removes the **budget, quota and cost tags** created in Notebook 04
3. 🛠️ Drops the **database, warehouse and roles** - everything the lab created
4. 🛠️ **Verifies** the account is clean and the shared CoWork object is intact

### Why it matters
A workshop that leaves objects behind costs money and confuses the next team. Cleanup is the last lesson in the lifecycle: remove only what you created, and prove it.

**Estimated time:** 5 min  |  **Prerequisite:** Any of Notebooks 00-05 (run this when you are finished)

> ⚠️ This notebook removes **only** the `STUDENT360_LAB` objects. Do not run it in an account where you need the Student 360 data afterwards.

```sql
%%sql -r set_context
USE ROLE ACCOUNTADMIN;
SELECT CURRENT_ROLE();
```

## Step 1: Unpublish the Agents from CoWork

🛠️ Only the lab's agents are removed. The shared CoWork object itself, and any agents other teams have published to it, are untouched.

```sql
%%sql -r unpublish
USE ROLE ACCOUNTADMIN;
ALTER SNOWFLAKE INTELLIGENCE SNOWFLAKE_INTELLIGENCE_OBJECT_DEFAULT DROP AGENT STUDENT360_LAB.AGENTS.ADVISOR_ASSISTANT;
ALTER SNOWFLAKE INTELLIGENCE SNOWFLAKE_INTELLIGENCE_OBJECT_DEFAULT DROP AGENT STUDENT360_LAB.AGENTS.STUDENT_SUCCESS_ANALYST;
```

## Step 2: Remove the Persona Grant from Your User

```sql
%%sql -r revoke_me
USE ROLE ACCOUNTADMIN;
SET ME = CURRENT_USER();
REVOKE ROLE STUDENT360_LAB_ADVISOR_ENGINEERING FROM USER IDENTIFIER($ME);
REVOKE ROLE STUDENT360_LAB_IR_ANALYST  FROM USER IDENTIFIER($ME);
```

## Step 3: Drop the Budget, Quota and Tags

```sql
%%sql -r drop_cost_objects
USE ROLE STUDENT360_LAB_ADMIN;
DROP SNOWFLAKE.CORE.BUDGET IF EXISTS STUDENT360_LAB.AGENTS.AI_BUDGET;
USE ROLE ACCOUNTADMIN;
DROP SNOWFLAKE.CORE.QUOTA IF EXISTS STUDENT360_LAB.AGENTS.AI_USER_QUOTA;
```

## Step 4: Drop the Database, Warehouse and Roles

🔹 Dropping the database removes the agents, semantic view, search services, dynamic tables, policies, eval dataset and stage in one step.

```sql
%%sql -r drop_database
USE ROLE ACCOUNTADMIN;
DROP DATABASE IF EXISTS STUDENT360_LAB;
DROP WAREHOUSE IF EXISTS STUDENT360_LAB_WH;
```

```sql
%%sql -r drop_roles
USE ROLE ACCOUNTADMIN;
DROP ROLE IF EXISTS STUDENT360_LAB_ADVISOR_ENGINEERING;
DROP ROLE IF EXISTS STUDENT360_LAB_IR_ANALYST;
DROP ROLE IF EXISTS STUDENT360_LAB_ADMIN;
```

## Step 5: Verify a Clean Account

```sql
%%sql -r verify_clean
USE ROLE ACCOUNTADMIN;
SHOW DATABASES LIKE 'STUDENT360_LAB';
SHOW WAREHOUSES LIKE 'STUDENT360_LAB%';
SHOW ROLES LIKE 'STUDENT360_LAB%';
SHOW AGENTS IN SNOWFLAKE INTELLIGENCE SNOWFLAKE_INTELLIGENCE_OBJECT_DEFAULT;
```

📌 **Expected:** the first three commands return no rows. The last command still lists any agents other teams published to the shared CoWork object - that is correct, and it is exactly why we only removed our own.

## ✅ Notebook 06 complete

### What you just did
- Removed the lab's agents from the shared CoWork object
- Dropped the budget, quota, tags, database, warehouse and roles
- Verified that nothing from the lab remains and the shared object is intact

---
### Key points
- Clean up is part of the lifecycle, not an afterthought.
- Remove only what you created: shared objects stay shared.
- Drop the database and everything under it goes with it, which is why one namespace per lab matters.
